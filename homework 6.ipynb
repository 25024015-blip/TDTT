W6A20
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
def kiem_tra(n,k):
    for i in range(len(n)):
        if n[i]==k:
            return i 
    return -1 
n=list(map(int,input().split()))
k=int(input())
print(kiem_tra(n,k))

    
```

W6A21
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi ra các tuple mà các ánh xạ đó xuất hiện


```python
# bài này tức là tìm vị trí mà các chuỗi đó xuất hiện
data=input().split()
kq={}
for i in range(len(data)):
    world=data[i]
    if world not in kq:
        kq[world]=[]
    kq[world].append(i)
for world in kq:
    kq[world]=tuple(kq[world])
print(kq)            
```

W6A27
Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước).
nếu không có in ra -1 

```python
def kiem_tra(data,x):
    kq=[]
    for i in data:
        if i<x:
            kq.append(i) 
    if len(kq)==0:
        return -1        
    return kq
data=list(map(int,input().split()))
x=int(input())
print(kiem_tra(data,x))
```

W6A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm


```python
data=list(map(int,input().split()))
so_cap_so=len(data)-1 
so_lan_tang=0 
so_lan_giam=0
for i in range(len(data)-1):
   if data[i] < data[i+1]:
      so_lan_tang+=1 
   else:
      so_lan_giam+=1 
if so_lan_tang==so_cap_so  :
   print("tăng")
elif so_lan_giam==so_cap_so:
   print("giảm")
else:
   print("không hợp lệ")             

            
```
