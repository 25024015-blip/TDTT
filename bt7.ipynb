Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu 
mỗi phần tử x trong mảng xuất hiện đúng x lần. 
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa 
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp. 

```python
data=list(map(int,input().split()))
danh_sach={}
for so in data: 
    if so in danh_sach:
        danh_sach[so]+=1
    else:
        danh_sach[so]=1 
so_luong_xoa= 0 
for x,count in danh_sach.items():
    if x<count:
        so_luong_xoa+=count-x
    elif x>count:
        so_luong_xoa+=count 
print(so_luong_xoa)                
```

Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng 
dần, lẻ giảm dần; ; 

```python
data=list(map(int,input().split())) 
kqc=[]
kql=[] 
for i in data: 
    if i%2==0:
        kqc.append(i)  
        kqc1=sorted(kqc)
    else: 
        kql.append(i)
        kql1=sorted(kql,reverse=True) 
if len(kqc)==0 :
    print(*kql1)
elif len(kql)==0:
    print(*kqc1)
elif len(kqc)!=0 and len(kql)!=0:
    print(*kqc1+kql1)        


            
```

W7A21 
Cho N là ngày, T là tháng năm 2020, Tính xem đó là ngày thứ bao nhiêu 
của năm đó nếu đó là ngảy hợp lệ 

```python
# Hàm phụ: Trả về số ngày của các tháng trong năm 2020 (Năm nhuận)
def so_ngay_trong_thang(thang):
    if thang in [1, 3, 5, 7, 8, 10, 12]:
        return 31
    if thang in [4, 6, 9, 11]:
        return 30
    if thang == 2:
        return 29  # Năm 2020 là năm nhuận!
    return 0

# --- HÀM ĐỆ QUY CHÍNH ---
def Kiem_tra(N, T):
    # Điều kiện dừng: Nếu là tháng 1 thì số ngày chính là N
    if T == 1:
        return N
    
    # Bước đệ quy: Lấy số ngày tháng trước + gọi đệ quy lùi về các tháng trước nữa
    return so_ngay_trong_thang(T - 1) + Kiem_tra(N, T - 1)


# --- CHƯƠNG TRÌNH CHÍNH ---
N = int(input("Nhập ngày N: "))
T = int(input("Nhập tháng T: "))

# Bước 1: Kiểm tra xem ngày tháng nhập vào có hợp lệ không
if T < 1 or T > 12 or N < 1 or N > so_ngay_trong_thang(T):
    print("Ngày không hợp lệ")
else:
    # Bước 2: Nếu hợp lệ, gọi hàm đệ quy để tính toán
    ket_qua = Kiem_tra(N, T)
    print(ket_qua)
    

    



    
```

W7A24 Cho xâu kí tự, xóa các từ giống nhau liền kề, chỉ giữ lại từ đầu tiên 

```python
data=input().split()
n=len(data)   
kq=[] 
for k in data: 
    kq.append(k)
for i in range(n-2,-1,-1): # thay vì quét xuôi thì ta quét ngược để dùng pop
    if data[i]==data[i+1]: 
        kq.pop(i+1)
print(*kq)       


         
```

W7A25 
Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi 
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số 

```python
n=input() # dùng thuật toán spling window
k=int(input())   
h=len(n)
for i in range(h-k+1): 
    mang=n[i:i+k] 
    maxx=int(mang) 
    for j in range(h-k+1):
        mangs=n[j:j+k]
        if int(mangs)>int(maxx):
            maxx=mangs 
print(maxx)            
```
